```python
board = ['_'] * 9
wins = [
    [0, 1, 2], [3, 4, 5], [6, 7, 8],
    [0, 3, 6], [1, 4, 7], [2, 5, 8],
    [0, 4, 8], [2, 4, 6]
]

def print_board():
    print()
    print(board[0], board[1], board[2])
    print(board[3], board[4], board[5])
    print(board[6], board[7], board[8])
    print()


def check_win(player):
    for combo in wins:
        if board[combo[0]] == board[combo[1]] == board[combo[2]] == player:
            return True
    return False

player = 'X'
print("Positions are numbered 0 to 8, left to right, top to bottom.")
print_board()

for turn in range(15):
    pos = int(input(f"Player {player}, enter position (0-8): "))

    if board[pos] != '_':
        print("That position is already taken. Try again.")
        continue

    board[pos] = player
    print_board()

    if check_win(player):
        print(f"Player {player} wins!")
        break

    if '_' not in board:
        print("It's a draw!")
        break

    if player == 'X':
        player = 'O'
    else:
        player = 'X'
```
